```python
import matplotlib.pyplot as plt
import numpy as np

# Data
request_rates = [0.5, 2, 5, 10]

# Deep/Nospec/Daspec data
deep_data = [359.22, 1216.76, 2179.07, 2880.82]
nospec_data = [357.78, 1200.76, 2170.06, 2907.72]
daspec_data = [356.35, 1230.90, 2243.06, 2992.29]

# Sharegpt data
sharegpt_daspec = [328.61, 1272.99, 2799.59, 4327.32]
sharegpt_smart = [328.67, 1243.72, 2437.80, 2599.85]
sharegpt_no = [328.63, 1256.68, 2758.98, 4441.85]

# Alpaca data
alpaca_daspec = [53.62, 214.14, 524.39, 1000.50]
alpaca_smart = [53.58, 214.36, 516.75, 994.88]
alpaca_no = [53.58, 214.37, 519.15, 984.47]

# Create figure with 3 subplots
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

# Plot Deep/Nospec/Daspec
x = np.arange(len(request_rates))
width = 0.25
ax1.bar(x - width, deep_data, width, label='Deep')
ax1.bar(x, nospec_data, width, label='Nospec')
ax1.bar(x + width, daspec_data, width, label='Daspec')
ax1.set_title('Sharegpt 100 reuqests')
ax1.set_xticks(x)
ax1.set_xticklabels(request_rates)
ax1.set_xlabel('Request Rate')
ax1.set_ylabel('Throughput (tok/s)')
ax1.legend()

# Plot Sharegpt
ax2.bar(x, sharegpt_smart, width, label='Smart')
ax2.bar(x - width, sharegpt_no, width, label='No')
ax2.bar(x + width, sharegpt_daspec, width, label='Daspec')
ax2.set_title('Sharegpt 300 requests')
ax2.set_xticks(x)
ax2.set_xticklabels(request_rates)
ax2.set_xlabel('Request Rate')
ax2.set_ylabel('Throughput (tok/s) ')
ax2.legend()

# Plot Alpaca
ax3.bar(x, alpaca_smart, width, label='Smart')
ax3.bar(x - width, alpaca_no, width, label='No')
ax3.bar(x + width, alpaca_daspec, width, label='Daspec')
ax3.set_title('Alpaca')
ax3.set_xticks(x)
ax3.set_xticklabels(request_rates)
ax3.set_xlabel('Request Rate')
ax3.set_ylabel('Throughput (tok/s)')
ax3.legend()

plt.tight_layout()
plt.savefig('benchmark_results.png')
plt.close()
```

```python

# Calculate deltas
# Deep/Nospec/Daspec deltas
deep_delta = [d - n for d, n in zip(deep_data, nospec_data)]
daspec_delta = [d - n for d, n in zip(daspec_data, nospec_data)]

# Sharegpt deltas
sharegpt_smart_delta = [s - n for s, n in zip(sharegpt_smart, sharegpt_no)]
sharegpt_daspec_delta = [d - n for d, n in zip(sharegpt_daspec, sharegpt_no)]

# Alpaca deltas
alpaca_smart_delta = [s - n for s, n in zip(alpaca_smart, alpaca_no)]
alpaca_daspec_delta = [d - n for d, n in zip(alpaca_daspec, alpaca_no)]

# Create figure with 3 subplots
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(15, 5))

# Plot Deep/Nospec/Daspec deltas
x = np.arange(len(request_rates))
width = 0.25
ax1.bar(x - width/2, deep_delta, width, label='Deep - Nospec')
ax1.bar(x + width/2, daspec_delta, width, label='Daspec - Nospec')
ax1.set_title('Sharegpt 100 requests')
ax1.set_xticks(x)
ax1.set_xticklabels(request_rates)
ax1.set_xlabel('Request Rate')
ax1.set_ylabel('Delta Throughput (tok/s)')
ax1.legend()
ax1.axhline(y=0, color='black', linestyle='-', alpha=0.3)

# Plot Sharegpt deltas
ax2.bar(x - width/2, sharegpt_smart_delta, width, label='Smart - No')
ax2.bar(x + width/2, sharegpt_daspec_delta, width, label='Daspec - No')
ax2.set_title('Sharegpt 300 requests')
ax2.set_xticks(x)
ax2.set_xticklabels(request_rates)
ax2.set_xlabel('Request Rate')
ax2.set_ylabel('Delta Throughput (tok/s)')
ax2.legend()
ax2.axhline(y=0, color='black', linestyle='-', alpha=0.3)

# Plot Alpaca deltas
ax3.bar(x - width/2, alpaca_smart_delta, width, label='Smart - No')
ax3.bar(x + width/2, alpaca_daspec_delta, width, label='Daspec - No')
ax3.set_title('Alpaca')
ax3.set_xticks(x)
ax3.set_xticklabels(request_rates)
ax3.set_xlabel('Request Rate')
ax3.set_ylabel('Delta Throughput (tok/s)')
ax3.legend()
ax3.axhline(y=0, color='black', linestyle='-', alpha=0.3)

plt.tight_layout()
plt.savefig('benchmark_delta_results.png')
plt.close()
```

```python
import pandas as pd

data = pd.read_csv('/data/AzureLLMInferenceTrace_conv_1week.csv')
```

```python
datas = data[:500]
# 确保TIMESTAMP是datetime格式
datas['TIMESTAMP'] = pd.to_datetime(datas['TIMESTAMP'])

# 创建一个时间序列，按秒分组并计算每秒的请求数
# 使用时间戳的日期部分+时间部分(精确到秒)作为新的索引
datas['second'] = datas['TIMESTAMP'].dt.floor('S')
qps = datas.groupby('second').size()

# 画出QPS图
import matplotlib.pyplot as plt
plt.figure(figsize=(15, 6))
qps.plot()

plt.xlabel('Time')
plt.ylabel('qps')
plt.grid(True)
plt.tight_layout()
plt.show()
```
